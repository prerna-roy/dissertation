## Zero-shot Prompts (Baseline) Development
### Mental Health Condition and Developmental Disability - Flags and Labels

To begin, baseline prompts will be created to extract the Mental Health Flag and Developmental Disability Flag variables, as well as their labels. 

Following the guidance from: https://arxiv.org/pdf/2309.13205 

### Version 0.1 (test)
Persona: You are an expert legal research assistant  
Context: extracting structured information from coronial documents.
Format: JSON output, flags must be "Yes", "No", labels can be free text (with multiple labels allowed)
Audience: For a public health researcher who requires a structured dataset
Tone: -- 

Prompt: You are an expert legal research assistant extracting structured information from coronial documents. The report can be accessed here: <URL>. Read the report and extract whether or not the decedent had a mental health condition and/or a developmental disability. If they did, record the result as "Yes", and if they did not, record it as "No". If they did, note the label of the mental health condition and/or developmental disability. Store the results in a JSON-format output with the variables "mental_health_flag", "developmental_disability_flag", "mental_health_label", and "developmental_disability_label".
